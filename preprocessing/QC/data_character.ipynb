```python
# Path to the data folder
folder_path = 'D:\\code\\uom_explore\\raw_data\\2024_08_29\\exp_brujin_seq_1'
# raw_data/2024_08_29/exp_brujin_seq_1
# D:\code\uom_explore\raw_data\2024_07_29\exp_efficiency_test_async_2
import os

# List to store the file names
file_names = []

# Iterate over all the files in the folder
for file in os.listdir(folder_path):
    # Check if the file is a file (not a directory)
    if os.path.isfile(os.path.join(folder_path, file)):
        file_names.append(file)

# Print the list of file names
print(file_names)
```

```python
import pandas as pd
import plotly.graph_objects as go
import os
from datetime import datetime

def parse_timestamp(file_name):
    try:
        # Extract the timestamp part from the file name
        timestamp_str = file_name.split('s')[0]
        # Parse the timestamp
        return datetime.strptime(timestamp_str, '%Y%m%d%H%M%S')
    except ValueError:
        print(f"Warning: Unable to parse timestamp from file name: {file_name}")
        return None

def read_sensor_value(file_path):
    try:
        # Use on_bad_lines='skip' instead of error_bad_lines
        df = pd.read_csv(file_path, on_bad_lines='skip')
        if 'sensor_value' in df.columns:
            return df['sensor_value'].iloc[0]
        elif len(df.columns) >= 3:
            return df.iloc[0, 2]
        else:
            print(f"Warning: Expected columns not found in file: {file_path}")
            return None
    except Exception as e:
        print(f"Error reading file {file_path}: {str(e)}")
        return None

# Extract timestamps and sensor values
data = []

for file in file_names:
    if '_BME680' in file:
        continue
    
    timestamp = parse_timestamp(file)
    if timestamp is None:
        continue
    
    file_path = os.path.join(folder_path, file)
    sensor_value = read_sensor_value(file_path)
    if sensor_value is not None:
        data.append((timestamp, sensor_value))

# Sort data by timestamp
data.sort(key=lambda x: x[0])

# Separate timestamps and sensor values
timestamps, sensor_values = zip(*data)

# Create the Plotly figure
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=timestamps,
    y=sensor_values,
    mode='lines+markers',
    name='Sensor Values'
))

fig.update_layout(
    title='Sensor Value vs Timestamp',
    xaxis_title='Timestamp',
    yaxis_title='Sensor Value',
    hovermode='x unified'
)

# Show the plot
fig.show()

# Print some statistics
print(f"Number of data points: {len(data)}")
print(f"Time range: {min(timestamps)} to {max(timestamps)}")
print(f"Sensor value range: {min(sensor_values)} to {max(sensor_values)}")
```

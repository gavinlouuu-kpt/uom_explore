```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('df.csv')

# Get all unique settings
all_settings = sorted(set([col.split('_')[-1] for col in df.columns if '_' in col]))
print(all_settings)
```

```python
# Select the settings you want to include
# selected_settings = ['140', '150', '152', '155', '157', '160', '162', '165', '167', '170', '172', '175', '177', '180', '182', '185', '187', '190', '192', '195', '197', '200', '202', '205', '210', '212', '215', '217', '220', '222', '225', '227', '230', '232', '235', '237', '240', '242', '245', '247', '250']  # Example: selecting only 140 and 150

selected_settings = ['140', '150', '152', '155', '157', '160', '162', '165', '167', '170', '172', '175', '177', '180', '182', '185', '187', '190', '192', '195']


# Create a new DataFrame with only the selected settings
new_df = pd.DataFrame()

# Add channel_id column
new_df['channel_id'] = df['channel_id']

# Add columns for selected settings
for setting in selected_settings:
    columns_to_add = [col for col in df.columns if col.endswith(f'_{setting}')]
    new_df = pd.concat([new_df, df[columns_to_add]], axis=1)

# Save the new DataFrame to a CSV file
new_df.to_csv('reduced/20_140_195_df.csv', index=False)

print(new_df.head())
print(f"Original columns: {len(df.columns)}")
print(f"New columns: {len(new_df.columns)}")
print(f"Selected settings: {selected_settings}")
```

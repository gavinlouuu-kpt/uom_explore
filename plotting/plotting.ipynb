```python
import pandas as pd

input = 'small_mlp.csv'
df = pd.read_csv(input)

# drop second and forth column
df.drop(df.columns[[1, 3]], axis=1, inplace=True)
# rename columns to epoch, loss, val_loss
df.columns = ['epoch', 'loss', 'val_loss']
df.head()
```

```python
# plot epoch vs loss and val_loss with plotly on 2 graph
import plotly.express as px
# set style as seaborn
# set word as bold in arial font


px.defaults.template = 'seaborn'
fig = px.line(df, x='epoch', y='loss', title='epoch vs loss')
fig.update_traces(line=dict(color='red'))
fig.show()

# set line color as red

fig = px.line(df, x='epoch', y='val_loss', title='epoch vs val_loss')
fig.update_traces(line=dict(color='blue'))
fig.show()
```
